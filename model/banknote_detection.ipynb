# Download Dataset

```python
from roboflow import Roboflow

rf = Roboflow(api_key="----")
project = rf.workspace("banknote-detection").project(
    "banknote_detection_yolo11-2-i4do6"
)
version = project.version(2)
dataset = version.download("yolov11", location="/content/drive/MyDrive/Capstone")
```

# Setup base model YOLOv11

```python
from ultralytics import YOLO

model = YOLO("yolo11s.pt")
```

# Training Model

```python
data_path = "/content/drive/MyDrive/Capstone/banknote_detection_yolo11-2/data.yaml"
results = model.train(
    data=data_path,
    epochs=10,
    batch=64,
    imgsz=640,
    workers=8,
    optimizer="AdamW",
    seed=77,
    project="/content/drive/MyDrive/Capstone",
    name="banknote_detection_models",
)
```

# Test Model

```python
custom_model = YOLO("/content/drive/MyDrive/Capstone/runs/detect/train/weights/best.pt")
```

```python
results = custom_model.predict(
    source="/content/drive/MyDrive/Capstone/banknote_detection_yolo11-2/test/images"
)
```

```python
import json

results = custom_model.predict(
    source="/content/drive/MyDrive/Capstone/New Images/IMG_20231013_173423_jpg.rf.be4324ab75be131c99b854ceb4930516.jpg",
    conf=0.7,
)

detections_list = []

for result in results:
    for box in result.boxes:

        x1, y1, x2, y2 = box.xyxy[0].tolist()

        width = x2 - x1
        height = y2 - y1

        score = float(box.conf[0])
        cls_id = int(box.cls[0])
        category_name = result.names[cls_id]

        detection = {
            "category_name": category_name,
            "score": score,
            "x_min": x1,
            "y_min": y1,
            "width": width,
            "height": height,
        }

        detections_list.append(detection)

final_json = {"detections": detections_list}

json_output = json.dumps(final_json, indent=4)
print(json_output)
```
